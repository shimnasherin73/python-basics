TWO DIMENSIONAL ARRAY:

```python
b=np.array([[1,2,3,4],[5,6,7,8,]])#2rows and 4 column
print(b)
```

```python
#indexing in 2d array:
#syntax:arr[row_index:column_index]
print(b[0,0]) #first row,first column
print(b[1,2])
```

```python
#access entire column:
print(b[:,2])#all rows,third column
```

```python
print(b[0])
```

```python
print(b[0:])#from zero to all
```

```python
print(b[0:2])#from zero to 2
```

```python
c=np.array([[3,8,-1,0],[4,8,1,1],[2,0,10,4],[1,2,3,4]])
print(c)
```

```python
#slicing sub array:
print(c[0:2,1:3])
```

```python
print(c[2,0:2])
```

```python
print(c[1:3,3])
```

```python
print(c[:,:1])
```

```python
print(c[1:,2])
```

ARRAY INITAILIZATION:

```python
#creating a array full of zeros
#default d type is float
a=np.zeros((3,4))#3rows and 4 columns
print(a)
```

```python
a.dtype
```

```python
#creating an array full of one 
np.ones((3,3))
```

```python
#specifying dtype
np.zeros((2,2),dtype=int) 
```

```python
np.ones((3,3),dtype=bool)
```

```python
#generate sequence of integers
# syntax:array_names=np.arrange(start,stop,step)
a=np.arange(1,10,2)
a
```

```python
b=np.arange(10,0,-2)
b
```

```python
#array name=np.full((row,column),default value)
c=np.full((3,3),"a")
c
```

```python
d=np.zeros((3,3))
d[[0,1,2],[0,1,2]]=1#after ,its column
d
```

```python
#generate a sequence of equally spaced values between start and end
#syntax:array name=np.linespace(start,stop,no of elements)
a=np.linspace(1,10,15)
a
```

```python
d=np.zeros((3,6))
d[[0,1,2],[0,0,0]]=1
d[[0,1,2],[2,2,2]]=3
d[[0,1,2],[4,4,4]]=5
d
```

```python
s=np.zeros((3,6))
s[:,[0,2,4]]=[1,3,5]
s
```

```python
k=np.zeros((3,6))
k[[0,1,2],[0,2,4]]=[1,3,5]
k
```

```python
s=np.zeros((3,6))
s[[0],[0,2,4]]=1
s[[0],[1,3,5]]=2
s[[1],[0,2,4]]=3
s[[1],[1,3,5]]=4
s[[2],[0,2,4]]=5
s[[2],[1,3,5]]=6
s

```

## STATISTICS:
- it is the sceince of collecting ,analyzing,interpreting and presenting data  
- stati help us to understnd data and make better decision 
- stati help to improve the accuracy of the ml model 


common concepts:
- mean
- mode
- median 
- variance
- standard deviation 

```python
#mean:(average of all values)
import numpy as np
marks=[70,80,90,100,60]
print("mean =",np.mean(marks))
```

```python
#median:(middle value of arranged data)
marks=[60,70,90,100,80]
print("median =",np.median(marks))
```

```python
#mode:(value that appear most frequntly )
from statistics import mode
mark=[40,60,90,80,70,70]
print("Mode =",mode(mark))
```

```python
#variance:(measures how far data values are spread from the mean )
marks=[70,80,90,100,60]
print("variance =",np.var(marks))
```

```python
#standard diviation:(square root of variance)
marks=[70,80,90,100,60]
print("standard diviation=",np.std(marks))
```

## Types of machine learning
1.SUPERVISED LEARNING :
here the models learns from labeled data.

- types of supervised learning:
  1.classification:cateogory /class
  2.regression :numeric value


1.classification :
 it is a supervised machine learning problem in which model predicts a categorical class label for the given data
- types of classification 
   1.binary classification :only have 2 classes 
   2.multi class classification:have more than 2 classes 
   3.multi-label classification :a single sample can belong to mmmultiple classes,eg:movie (horro,action etc..)
   4.imbalanced classification :classes have unequal number of samples ,one class have more samples than others

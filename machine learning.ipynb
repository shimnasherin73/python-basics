## STATISTICS:

common concepts:

```python
#mean:(average of all values)
import numpy as np
marks=[70,80,90,100,60]
print("mean =",np.mean(marks))
```

```python
#median:(middle value of arranged data)
marks=[60,70,90,100,80]
print("median =",np.median(marks))
```

```python
#mode:(value that appear most frequntly )
from statistics import mode
mark=[40,60,90,80,70,70]
print("Mode =",mode(mark))
```

```python
#variance:(measures how far data values are spread from the mean )
marks=[70,80,90,100,60]
print("variance =",np.var(marks))
```

```python
#standard diviation:(square root of variance)
marks=[70,80,90,100,60]
print("standard diviation=",np.std(marks))
```

## logistic regression:
it is a a supervised learning classification algorithm used to predict the probabilty of a cateogorical outcome 
- logistic regression is mainly used for classification rather than regression because it predict the probability of a class not continuos numerical value .
- predicted probabilty is then converted int to class label  (0,1)using threshold. 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

#load dataset:
iris=load_iris()
x=iris.data#(input features)
y=iris.target#(output labels)

#split data:
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#create model :
model=LogisticRegression(max_iter=200)

#train model:
model.fit(x_train,y_train)

#predict:
y_pred=model.predict(x_test)

#accuracy:
accuracy=accuracy_score(y_test,y_pred)

print("predictions:",y_pred)
print("accuracy:",accuracy)
```

## 3.knn(k nearest neibours):
it is a supervised classification algorithm that classifies the sample based on the majority class of  its nearset neibours 

how it works :
- choose the value of k
- find the k nearset data points 
- count the classes
- assign the majority classes

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("accuracy:",accuracy_score(y_test,y_pred))
```

how it works:
- calculate the probability of each class 
- calculate the probability of features of each class

what is the main assemption of naive byes?
- all input features are assumed to be independent of eachother 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=GaussianNB()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

## 6.Random forest:

it is a ensemble(collection)classification algorithm that combine multiple decision tree and predicts the final class using majority voting

how:
- create many decision tree
- each tree is trained on a random subset of data
- each tree gives prediction 
- the majority vote become the final prediction 


overfitting:
preform well during training time and perform less during the time of testing

underfitting:
perform less on both testing and traininig

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

## 7.gradient boosting:

it is an ensenmble classification algorithm that build decision tree sequentially.where each tree tries to correct the errors made by the previous tree

how:
- train the first decision tree
- calculate the errors
- train the next tree to reduce the errors 
- repeat the process 
- combine  all the trees for the final prediction
  

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble  import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=GradientBoostingClassifier(random_state=42)#(n estimate will be get by deafult 100)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

## Types of regression :

1.simple linear regression.
2.multiple linear regression.
3.polynomial regression.

1.simple linear regression.
it uses one independent variable to predict one dependent variable 
  - only one input feature is used for prediction

eg:-study hours-mark

```python
import numpy as np 
from sklearn.linear_model import LinearRegression

x=np.array([[1],[2],[3],[4]])
y=np.array([40,50,60,70])

model=LinearRegression()

model.fit(x,y)

print(model.predict([[5]]))
```

## 2.multiple linear regression :
multiple linear regression uses two or more independent variables to predict one dependent variable 

- multiple input features are used for prediction 


```python
#study hour+attendence=mark
import numpy as np
from sklearn.linear_model import LinearRegression

x=np.array([[2,70],[4,75],[6,80],[8,90]])
y=np.array([40,55,70,85])

model=LinearRegression()
model.fit(x,y)
print(model.predict([[5,80]]))
```

## 3.polynomial regression:

- it is used when the relationship between variables is non-linear. 
- it is a curved line instead of straight line.

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

x=np.array([[1],[2],[3],[4]])
y=np.array([2,5,10,17])

poly=PolynomialFeatures(degree=2)#degree2=square of x
x_poly=poly.fit_transform(x)#square of x wil be output 

model=LinearRegression()
model.fit(x_poly,y)#sqaure of x+ y(differcncr between square and y)

x_new=poly.transform([[5]])
print(model.predict(x_new))
```

## ROC AND AUC:-
ROC:- (receiver operating charecterstic)
- is a graph used to evaluate the performance of a binary classifiction model
- it shows well the model distingush between positive and negative classes
  
AUC:-
- AUC (area under roc curve) is the area under the roc curve
- it measure the overall performance of the classification model 
  

why do we use roc and auc:-
- to evalute binary classification models 
- to compare different classification models 
- to measure the models ability to seperte classes 

roc:
x -> false positive rate(FPR) 
y -> true positive rate(TPR)

```python
from sklearn.metrics import roc_curve,auc
import matplotlib.pyplot as plt

y_true=[0,0,1,1]
y_score=[0.1,0.4,0.35,0.8]

fpr,tpr,threshold=roc_curve(y_true,y_score)

roc_auc=auc(fpr,tpr)

print("AUC SCORE=",roc_auc)

plt.plot(fpr,tpr,label="ROC CURVE")
plt.plot([0,1],[0,1],"--")#(x,y axis line upto 0.1 to 1.0  )
plt.xlabel("False postive rate")
plt.ylabel("true positive rate")
plt.title("ROC CURVE")
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt 

#datasets:
x=[[1,2],[2,3],[3,3],[8,8],[9,8],[8,9]]

#create model:
model=KMeans(n_clusters=2,random_state=42)

#train:
model.fit(x)

#predict:
labels=model.labels_
print(labels)

#plot:
plt.scatter(
    [i[0]for i in x],
    [i[1]for i in x],
    c=labels,
    s=100)

#plt for centoird:
plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker="X",
    s=200
)
plt.title("K-means clustering ")
plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

#load data:-
iris=load_iris()

x=iris.data

#standerdize :
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)

#apply pca :
pca=PCA(n_components=2)
x_pca=pca.fit_transform(x_scaled)

print("original shape:",x.shape)
print("pca shape:",x_pca.shape)
```

VISUALISING PCA:

```python
import matplotlib.pyplot as plt
plt.scatter(x_pca[:,0],x_pca[:,1])
plt.xlabel("pricipal component 1")
plt.ylabel("pricipal component 2")
plt.title("PCA VISUALIZATION")
plt.show()
```

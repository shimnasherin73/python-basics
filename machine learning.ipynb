```python
#mean:(average of all values)
import numpy as np
marks=[70,80,90,100,60]
print("mean =",np.mean(marks))
```

```python
#median:(middle value of arranged data)
marks=[60,70,90,100,80]
print("median =",np.median(marks))
```

```python
#variance:(measures how far data values are spread from the mean )
marks=[70,80,90,100,60]
print("variance =",np.var(marks))
```

```python
#standard diviation:(square root of variance)
marks=[70,80,90,100,60]
print("standard diviation=",np.std(marks))
```

## logistic regression:
it is a a supervised learning classification algorithm used to predict the probabilty of a cateogorical outcome 
- logistic regression is mainly used for classification rather than regression because it predict the probability of a class not continuos numerical value .
- predicted probabilty is then converted int to class label  (0,1)using threshold. 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

#load dataset:
iris=load_iris()
x=iris.data#(input features)
y=iris.target#(output labels)

#split data:
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#create model :
model=LogisticRegression(max_iter=200)

#train model:
model.fit(x_train,y_train)

#predict:
y_pred=model.predict(x_test)

#accuracy:
accuracy=accuracy_score(y_test,y_pred)

print("predictions:",y_pred)
print("accuracy:",accuracy)
```

## 3.knn(k nearest neibours):
it is a supervised classification algorithm that classifies the sample based on the majority class of  its nearset neibours 

how it works :
- choose the value of k
- find the k nearset data points 
- count the classes
- assign the majority classes

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("accuracy:",accuracy_score(y_test,y_pred))
```

how it works:
- calculate the probability of each class 
- calculate the probability of features of each class

what is the main assemption of naive byes?
- all input features are assumed to be independent of eachother

## 6.Random forest:

it is a ensemble(collection)classification algorithm that combine multiple decision tree and predicts the final class using majority voting

how:
- create many decision tree
- each tree is trained on arandom subset of data
- each tree givves prediction 
- the majority vote become the final prediction 


overfitting:
preform well during training time and perform less during the time of testing

underfitting:
perform less on both testing and traininig

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

## 7.gradient boosting:

it is an ensenmble classification algorithm that build decision tree sequentially.where each tree tries to correct the errors made by the previous tree

how:
- train the first decision tree
- calculate the errors
- train the next tree to reduce the errors 
- repeat the process 
- combine  all the trees for the final prediction
  

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble  import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=GradientBoostingClassifier(random_state=42)#(n estimate will be get by deafult 100)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

## Types of regression :

1.simple linear regression:
2.multiple linear regression
3.polynomial regression

1.simple linear regression
it uses one independent variable to predict one dependent variable 
- only one input feature is used for prediction

eg:-study hours-mark

```python
import numpy as np 
from sklearn.linear_model import LinearRegression

x=np.array([[1],[2],[3],[4]])
y=np.array([40,50,60,70])
model=LinearRegression()
model.fit(x,y)
print(model.predict([[5]]))
```

## 2.multiple linear regression :
multiple linear regression uses two or more independent variables to predict one dependent variable 

- multiple input features are used for prediction 


```python
#study hour+attendence=mark
import numpy as np
from sklearn.linear_model import LinearRegression

x=np.array([[2,70],[4,75],[6,80],[8,90]])
y=np.array([40,55,70,85])
model=LinearRegression()
model.fit(x,y)
print(model.predict([[5,80]]))
```

## 3.polynomial regression:

- it is used when the relationship between variables is non-linear
- it is a curved line instead of straight line 

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

x=np.array([[1],[2],[3],[4]])
y=np.array([2,5,10,17])

poly=PolynomialFeatures(degree=2)#degree2=square of x
x_poly=poly.fit_transform(x)#square of x wil be output 

model=LinearRegression()
model.fit(x_poly,y)#sqaure of x+ y(differcncr between square and y)

x_new=poly.transform([[5]])
print(model.predict(x_new))
```

```python
import numpy as np 
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

x= np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 6, 11, 18, 27])

poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)

model=LinearRegression()
model.fit(x_poly,y)

x_new=poly.transform([[6]])
print(model.predict(x_new))

```

1.linear regression:
is a supervised learning algorithm used to predict continuos numerical value
- it models the relationship between the input variable and output variable using straight line 

equation:
- y=mx+b

adventages:
- simple and easy to understand
- fast to train
- easy to interprets

disadventages:
- assume a linear relationship
- sensitive to outliers
- not suitable for  complex non linear data


```python
from sklearn.linear_model import LinearRegression
import numpy as np 
import matplotlib.pyplot as plt 

# training data
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

#create model
model=LinearRegression()

#train model
model.fit(x,y)

#predict
y_pred=model.predict(x)

#predict new value
print(model.predict([[6]]))

#plotting:
plt.scatter(x,y)
plt.plot(x,y_pred,color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.title("linear regression")
plt.show()
```

## feature scaling method:
it is a process of bringing all numerical features to similar range
- it helps machine learning algorithm to perform better and converge faster
- converge:reduce error and improving performance 
  
-  2 methods:
1. normalization:normalization scales the value between 0-1(min-max scaling)
2. standarization:it transform the data to have a mean of 0 and standard deviation of 1.after standardization,the data is centered around 0

when to use;
1. normalization:when the data has fixed range
2. standardization:when the data contain outliers or follow a normal distribution 

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
import numpy as np

data=np.array([[10],[20],[30],[40],[50]])

#normalization:
minmax=MinMaxScaler()
print("noramalization:\n",minmax.fit_transform(data))

#standardization:
standard=StandardScaler()
print("standardization:\n",standard.fit_transform(data))
```

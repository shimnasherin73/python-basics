## STATISTICS:
- it is the science of collecting ,analyzing,interpreting and presenting data  
- stati help us to understnd data and make better decision 
- stati help to improve the accuracy of the ml model 


common concepts:
- mean
- mode
- median 
- variance
- standard deviation 

```python
#mean:(average of all values)
import numpy as np
marks=[70,80,90,100,60]
print("mean =",np.mean(marks))
```

```python
#median:(middle value of arranged data)
marks=[60,70,90,100,80]
print("median =",np.median(marks))
```

```python
#mode:(value that appear most frequntly )
from statistics import mode
mark=[40,60,90,80,70,70]
print("Mode =",mode(mark))
```

```python
#variance:(measures how far data values are spread from the mean )
marks=[70,80,90,100,60]
print("variance =",np.var(marks))
```

```python
#standard diviation:(square root of variance)
marks=[70,80,90,100,60]
print("standard diviation=",np.std(marks))
```

1.classification :

- types of classification

   3.multi-label classification :a single sample can belong to multiple classes,eg:movie (horro,action etc..)

## logistic regression:
it is a a supervised learning classification algorithm used to predict the probabilty of a cateogorical outcome 
- logistic regression is mainly used for classification rather than regression because it predict the probability of a class not continuos numerical value .
- predicted probabilty is then converted int to class label  (0,1)using threshold. 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

#load dataset:
iris=load_iris()
x=iris.data#(input features)
y=iris.target#(output labels)

#split data:
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#create model :
model=LogisticRegression(max_iter=200)

#train model:
model.fit(x_train,y_train)

#predict:
y_pred=model.predict(x_test)

#accuracy:
accuracy=accuracy_score(y_test,y_pred)

print("predictions:",y_pred)
print("accuracy:",accuracy)
```

## 3.knn(k nearest neibours):
it is a supervised classification algorithm that classifies the sample based on the majority class of  its nearset neibours 

how it works :
- choose the value of k
- find the k nearset data points 
- count the classes
- assign the majority classes

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("accuracy:",accuracy_score(y_test,y_pred))
```

## 4.decision tree:
make decision by splitting data into branches based on feature values 

how it works:
- start from root node 
- ask a conditon
- split the data into branches
- continue splitting until a final class is reached
  

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=DecisionTreeClassifier(random_state=42)#randomstate to make it stable
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("accuracy:",accuracy_score(y_test,y_pred))
```

## 5.naive bayes
supervised classification algorithm based on bayes theorem that predict the class with highest probability 

how it works:
- calculate the probability of each class 
- calculate the probability of features of each class 
- choose the class with highest probability 



what is the main assemption of naive byes?
- all input features are assumed to be independent of eachother 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

iris=load_iris()
x_train,x_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=GaussianNB()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
```

```python
import numpy as np 
from statistics import mode
marks=[25,56,79,70,80,90,45,87]
print("mean=",np.mean(marks))
print("mode=",mode(marks))
print("median=",np.median(marks))
print("standard=",np.std(marks))
print("variance=",np.var(marks))
```

```python
#supervised learning:
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

iris=load_iris()
#x=iris.data
#y=iris.target

x_train,x_test,y_train,y_tests=train_test_split(iris.data,iris.target,test_size=0.2,random_state=42)
model=LogisticRegression(max_iter=200)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
accuracy=accuracy_score(y_pred,y_tests)
print("predictions:",y_pred)
print("accuracy:",accuracy)
```

```python
import numpy as np 
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

x=np.array([[1],[2],[3],[4]])
y=np.array([2,5,10,17])

poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)

model=LinearRegression()
model.fit(x_poly,y)

x_new=poly.transform([[5]])
print(model.predict(x_new))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

x,y=load_iris(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=RandomForestClassifier()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(classification_report (y_test,y_pred) )
```

```python
import matplotlib.pyplot as plt 
from sklearn.cluster import KMeans

x=[[1,2],[2,3],[3,3],[8,8],[9,8],[8,9]]
model=KMeans(n_clusters=2,random_state=42)
model.fit(x)
labels=model.labels_
print(labels)

plt.scatter(
    [i[0]for i in x],
    [i[1] for i in x],
    c=labels,
    s=100
)
plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker="x",s=200
)
plt.title("K-means clustering ")
plt.show()
```

1.LINE CHART:
a line chart display data points connected by straight line segments.perfect for showing trends over time or continuos data.

MAIN PARAMETERS:
 
 PARAMETERS        DESCRIPTION              EXAMPLE 
x,y              data points for x        plt.plot(x,y)
                 and y axis 
colour/c         line colour              colour='blue,or c='b'
linestyle/ls     line style(-,--,-.,:)    ls="--"
linewitdh/lw     line width in points     linewidth=2
marker           marker style (o,s,^,*)   marker=o
markersize/ms    size of markers          marker size=6
alpha            transparency(0-1)        alpha=0.7
label            legend label             label='series1'

```python
import matplotlib.pyplot as plt 
x=[1,2,3,4]
y=[10,20,15,25]

plt.plot(x,y)# plotting 
plt.show()
```

```python
import numpy as np 
x=np.linspace(0,10,50)# get random 50 numbers with equal range between 1-10
y=np.sin(x)

plt.figure(figsize=(8,5))#size of the chart 
plt.plot(x,y, color='blue',linewidth=2,linestyle=':',marker='o',markersize=5,label='sin(x)',alpha=0.8)
#plt.xlim(-1,10),sets the limit for the x axis from -1 to 10 
#plt.ylim(10,90),sets the limit for y axis from 10 to 90 

plt.xlabel('x Axis')
plt.ylabel('Y Axis')
plt.title('line chart example')
plt.legend()#if label is given ,to show the label 
plt.grid(True,alpha=0.5)# grid under the chart 
plt.show()
```

FORMAT STRING :

A format string is a convenient way to specify the color ,marker and lune style of plot in a single string 

color:you can specify the color using a single charecter string from the following options :
'b':blue
'g':green
'r':red
'c':cyan
'm':magneta
'y':yellow
'k':black
'w':white

Marker: this specify the type of marker to use for data points
'-':point
',':pixel
'o':circle
'v ' :traingle down
'^':traingle lup
'>':traingle left
'<':traingle right 
'1':tri down 
'2':tri up
'3':tri left
'4':tri right 
's':square 
'p':pentagon 
"*":star 
'h':hexagon1
"H":hexagon2
'+':plus
'x': x
'D':diamond 
'd':thin diamond 
'|':vertical line 
'_':horizonatl line 


line style :this specifies the style of the line connecting the data points :
'-':solid line 
'--':dashed line 
'-.':dash-dot line 
':':dotted lines 

COMBINING FORMAT STRING :

```python

import matplotlib.pyplot as plt 
x=['shaym','aswin','praveen','anjana','jubil','shibin']
y=[86,75,76,74,72,85]

plt.plot (x,y,'rs')
plt.title('person vs weight diagram')

plt.xlabel('Name')
plt.ylabel('Weight')
plt.show()
```

```python
x=['shaym','aswin','praveen','anjana','jubil','shibin']
y=[86,75,76,74,72,85]

plt.plot(x,y,'g--')
plt.title('person vs weight diagram')

plt.xlabel('Name')
plt.ylabel('Weight')
plt.show()
```

```python
x=['shaym','aswin','praveen','anjana','jubil','shibin']
y=[86,75,76,74,72,85]
 
plt.plot(x,y,'b*-',linewidth=2)
plt.title('person vs weight diagram')

plt.xlabel('name')
plt.ylabel('weight')
plt.show()
```

```python

#fmt='s--b':s stands for square markers ,--dashed lines and b for blue colour
fmt='s--b'
x_val=['shaym','aswin','praveen','anjana','jubil','shibin']
y_val=[86,75,76,74,72,85]

plt.plot(x_val,y_val,fmt)
plt.title('Person vs Weight diagram')

plt.xlabel('Name')
plt.ylabel('Weight')
plt.show()

```

MULTIPLE PLOT:

```python
x_val=['shaym','aswin','praveen','anjana','jubil','shibin']
y_val=[86,75,76,74,72,85]

plt.plot(x_val,y_val,label='current weight',c='red',linestyle='-.',marker='o')# sets the color of the line to red 

y_val2=[75,75,80,65,70,78]
plt.plot(x_val,y_val2,label='desired weight',c='green',linestyle=':',marker='s')#sets the marker style to square 

plt.title('person vs weight diagram')
plt.xlabel('Name')
plt.ylabel('Weight')
plt.legend()
plt.show()
```

```python
import numpy as np 
x=np.linspace(0,10,50)# get random 50 numbers with equal range between 1-10
y=np.sin(x)

plt.figure(figsize=(8,5))#size of the chart 
plt.plot(x,y, color='blue',linewidth=2,linestyle=':',marker='o',markersize=5,label='sin(x)',alpha=0.8)
#plt.xlim(-1,10),sets the limit for the x axis from -1 to 10 
#plt.ylim(10,90),sets the limit for y axis from 10 to 90 

plt.xlabel('x Axis')
plt.ylabel('Y Axis')
plt.title('line chart example')
plt.legend()#if label is given ,to show the label 
plt.grid(True,alpha=0.5)# grid under the chart 
plt.show()
```

SCATTER PLOT 

a scatter plot displays data points without connecting lines .useful for showing relationship between two variables 

main parameters 


parametrs              description               example 
x,y                    data cordinates           plt.scatter(x,y)
s                      marker size (area)        s=50
c                      marker color              c='red'
marker                 marker shape              marker="o"
alpha                  transparancy              alpha=0.6
edgecolors             edge color of markers     edge color='black'
cmap                   color map of color array   cmap='viridis'
vmin,vmax              data range for color map    vmin=0,vmax=100

```python
import matplotlib.pyplot as plt
import numpy as np  
plt.figure(figsize=(8,5))
x=np.linspace(0,10,50)
y=np.sin(x)
colors_array=np.random.rand(50)#create random numbers between 0-1.theses value decie the color of each point 
sizes=np.random.uniform(50,300,50)#create 50 random sizes from 50 -300.these shows how big the dot is 
plt.scatter(x,y,c=colors_array,cmap='viridis',alpha=0.6,edgecolors='black',linewidth=0.5)

#cmap chooses the color map . viridis convert the number in to colors
#alpha controls transparancy .range:0 completely invisible ,1:completely solid 

plt.colorbar(label='color values ')# add a color scale beside the graph .it shows what the color represent 
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('scatter plot with color and size mapping')
plt.grid(True,alpha=0.3)
plt.show()
```

PIE CHART 

a pie chart is a circular statistical grpah that is divided in to slices to illustrate numerical propotions 
each slice of the pie represent a categories contribution to whole ,making it easy to compare differnt partes of the data sets 

use a method pie()

```python
slices=[30,40,50]#slices represents men(30),women(40),kids(50)
var=['men','women','kids']
plt.style.use('dark_background')#style
plt.pie(slices,labels=var)#this function create the pie chart using the data and label provided 
plt.title('PIE CHART')
plt.show()
```

CUSTOM COLORS AND WEDGE PROPEERTIES :

WEDGE PROPERTIES:
in a pie chart ,a wedge or a (sector)is one of the slices that make up the chart 
each wedge represent the proportionate part of the whole data set 
the size of each wedge is determined by the value it represent relative to the total sum of all values 
you can customise the appearence of the pie chart wedges using wedgeprops parameters 

LINE WIDTH AND EDGE COLOR:
border line .
example:wedgeprops={'linewidth':2,edgecolor='yellow'}sets the line width to 2 and the edge color to yellow 

EXPLODS AND SHADOWS:
you can use the explode parameter to seperate wedges and shadows=True to add a shadow effect 

```python
slices=[30,40,50]
var=['men','women','kids']
col=["yellow","purple","pink"]

plt.pie(slices,labels=var,colors=col,wedgeprops={'linewidth':2,'edgecolor':'red'})#this function create the pie chart using the data and labels provided 
plt.title("PIE CHART")
plt.style.use('classic')
plt.show()
```

EXPLODES AND SHADOWS :

```python
slices=[30,40,20]
var=['men','women','kids']#provide label for each slices 
col=["yellow","purple","pink"]

explodes=[0,0,0.2]#explode the third slide 
plt.pie(slices,labels=var,colors=col,wedgeprops={'linewidth':2,'edgecolor':'red'},explode=explodes,shadow=True)#this function create the pie chart using the data and labels provided 
plt.title('PIE CHART')
plt.style.use('classic')
plt.show()
```

START ANGLE AND PERCENTAGE VALUE :

```python
slices=[30,40,20]
var=['men','women','kids']#provide label for each slices 
col=["red","blue","pink"]

explodes=[0,0,0.2]#explode the third slide 
plt.pie(slices,labels=var,colors=col,wedgeprops={'linewidth':2,'edgecolor':'black'},explode=explodes,shadow=True,startangle=90,autopct='%.2f%%')
#start angle=90 starting with 90 degree angle and autocpt=percentage and  how many digits have to show after points of percentage  
plt.title('PIE CHART')

plt.style.use('classic')
plt.show()

#explodes:list specifying offset of each segments 
#shadow:tah shadow =true parametr add a shadow to the pie chart 
#start angle :rotate the start of the pie chart ny 90 degrees 
#autocpt:automatic percentage.it is used to display percentage values on a pie chart slices ,formatted to the two decimal points  
```

STACK PLOT OR AREA CHART:
a stack plot in matplotlib is a type of plot that display multiple datasets as vertically stacked areas .
this is useful for visualising the cumulative  value of multiple data series over time and continous variable.
each data set is represented by a differnt color ,and tha area are stacked on the top of each other ,showing how each parts contribution 

```python
import matplotlib.pyplot as plt
days=[1,2,3,4,5]
studying=[6,8,7,6,11]
playing=[1,8,7,6,13]

plt.stackplot(days,studying,playing,colors=['lightblue','pink'])
plt.xlabel("days")
plt.ylabel('number of hours')
plt.legend(['studying','playing'],loc='upper left')
plt.show()
```

HISTOGRAM:

It is the graphical representation used to visualise the distribution of a dataset or frequency distribution.
it organise a continuos or discrete variable in to "bins" or intervals and display the frequency or counts of data points with in each bin .
The x-axis represents the range of values and the y-axis represents the frequency of data points in each range.

Bins: - these are the intervals into which the data range is divided. Each bin represents a range of values.

Frequency: - the height of each bar in the histogram represents the no.of data points that fall within that bin.

Continuos data: - histograms are used for continous data, where the data points can take any value within a range.

NORMAL DISTRIBUTION :-BELL SHAPED CURVE 

```python
import numpy as np 
import matplotlib.pyplot as plt

data=np.random.normal(40,2,1000)

plt.hist(data,bins=10,edgecolor='black',color='skyblue')
plt.xlabel("value in x-axis")
plt.ylabel("values in y label")
plt.title('histogram of normally distributed data')

plt.show()
```

UNIFORM DISRTIBUTION 
-all outcomes are equally likely within a given range .


```python
import matplotlib.pyplot as plt 
import numpy as np 

data=np.random.uniform(0,10,1000)

plt.hist(data,bins=20,color='pink',edgecolor='black')
plt.xlabel('values')
plt.ylabel('frequency')
plt.title('uniform distribtuion of histogram')
plt.show()
```

OUTLIER:
it is a data points that significantly deviates from other observation in the data set outliers can be usually high,or low compared to the majority of data

any data points that falls outside the whiskers (ie,beyond 1.5 IQR from Q1 or Q2)is considerd an outlier and its usually  marked with a dot or astertiks
 

BOX PLOT OR BOX -AND-WHISKER PLOT:
its the standerdised way displaying the distribution of data based on five number summary 
minimum ,first quartile(q1)median(q2),third quartile(q3) and maximum.
it provides a visual representation of the spread and skeweness of the data ,while aslo highlighting outliers 

KEY COMPONENTS OF BOX PLOT :
box:
the box itself  represent the IQR which is the range between the first quartile (Q1)and third quartile(Q3).
it contain the middle 50% of the data

the line inside the box:
this is median (Q2) or the middle value of the data

WHISKERS :-
the whiskers extend from the box to the minimum and maximum values that are not considerd as outliers .
whiskers typicaly cover the range witin 1.5 *IQR from Q1 and Q3 

USES OF THE BOXPLOT:
-visualise the spread and skewenes of the data 
-compare distribution between differnt data set
-identify outliers in the data 
-summerize large data set  in a compact form 

COMMON METHOD INCLUDE :
*z-score -identifies outliers by measuring how many standerd deviations a data point is from mean 

*IQR -inter quartile range -outliers are detected as values lying below 

*box plot-visualizes outliers as points outside the whiskers of the box plot

*scatter plot or histogram -visual detection of points 


```python
import numpy as np
import matplotlib.pyplot as plt

data=np.random.normal(0,1,100)
plt.boxplot(data)
plt.title('box plot')
plt.ylabel('values')
plt.show()
```

4.PLot a simple line graph of the function y=x2 for values of x from -10 to 10.Add Labels for the x-axis and y-axis, and a title.

7.Generate a scatter plot with random x and y data points. Color each point according to its distance from the origin.
Set the size of each point according to its value 

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.rand(50)*100
y=np.random.rand(50)*100

distance=np.sqrt(x**2+y**2)

plt.scatter(x,y,c=distance,s=50,cmap='viridis')
plt.title('scatter plot')
plt.xlabel('x')
plt.ylabel('y')

plt.colorbar(label='distance from origin')
plt.grid()
plt.show()
```

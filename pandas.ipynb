DATA TYPE USING PYTHON TUPLE LIST:

```python
data_tup=[(101,"sam",15,49),
          (102,"alan",16,39),
          (103,"jessy",17,44),
          (104,"kate",18,45),
          (105,"hardin",15,42)]
dataframe=pd.DataFrame(data_tup,columns=['id','name','age','marks'])
dataframe
```

LINEAR INTERPOLATION:

Linear interpolation is a technique used to estimate missing values 
in a dataframe or series by filling in the gaps with values that lie on the stright line between knwon data points 
specifically,you can use the interpolate ()method with the method ='linear'parameter
the method ='linear'is the default for the dataframes and it treats the index value as equally spaced 

```python
#creating datafrma ewith missing values:
import numpy as np
df=pd.DataFrame({'A':[1,2,np.nan,4],
              'B':[5,np.nan,np.nan,8],
              'C':[9,10,11,12]})

# applying linear interpolation
df_interpolated=df.interpolate(method='linear')
print(df_interpolated)
```
